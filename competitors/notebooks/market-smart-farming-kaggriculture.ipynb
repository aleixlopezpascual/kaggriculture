## 4. Freeze the exact multi-file submission

```python
import base64, hashlib, json, tarfile, py_compile
from pathlib import Path

VARIANT = "four_action_sale_reservation"
EXPECTED_ARCHIVE_SHA256 = "872d951cdde623d00f429245f62276d21df45d0cd82a84350f7e4931888c89c2"
EXPECTED_MEMBER_SHA256 = {
    "actions.json": "17d503f2fd20d59f9c0f14024d1e74a8add8bb9b5561d4d908b45deecb5495ef",
    "main.py": "495bfa4825c9e58638e804aeeb62a9537826f6d4d869222815e1bbeefd84d3dc",
    "policy.py": "79d5804ec9d94ad14fa8df72c6c73e2f81cd7dc0e5b2b4e77d7b50b59db08a16",
    "router.py": "3a444a70ea9dd89f3cb857676fcc0180ebf1dfb7ede89f7f0cc82a2ef384e380",
    "settings.json": "90c07e51dc1e62dde2fc76414ce471017a90a26efc95357ff5d958d618271892"
}
EXPECTED_MEMBERS = ['actions.json', 'main.py', 'policy.py', 'router.py', 'settings.json']
ARCHIVE_B64 = """"""

archive = Path("submission.tar.gz")
archive.write_bytes(base64.b64decode(ARCHIVE_B64))
assert hashlib.sha256(archive.read_bytes()).hexdigest() == EXPECTED_ARCHIVE_SHA256

with tarfile.open(archive, "r:gz") as tf:
    names = tf.getnames()
    assert names == EXPECTED_MEMBERS
    for name in names:
        data = tf.extractfile(name).read()
        assert hashlib.sha256(data).hexdigest() == EXPECTED_MEMBER_SHA256[name]
        Path(name).write_bytes(data)

for name in ("main.py", "policy.py", "router.py"):
    py_compile.compile(name, doraise=True)
print("runtime files:", ", ".join(EXPECTED_MEMBERS))
print("archive SHA-256:", EXPECTED_ARCHIVE_SHA256)
print("runtime reconstruction: PASS")
```
